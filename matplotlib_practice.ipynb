# Scatter Plots

```python
# plt.scatter on pandas data
df = pd.read_csv('data/batter.csv')
df = df.head(25)
df.head(10)
```

```python

plt.figure(figsize=(15, 7))
plt.scatter(df['avg'], df['strike_rate'], s=df['runs']/50, marker='o', cmap='jet', c=df['runs'])
plt.title('Strike Rate vs Avg')
plt.ylabel('Strike Rate')
plt.xlabel('Avg')

plt.colorbar()

plt.axhline(125, color='red')
plt.axvline(30, color='red')

for _, row in df.iterrows():
    plt.text(row['avg'], row['strike_rate'], f"{row['batter']}({row['runs']})", fontsize=8)
```

# Stacked Bar

```python
# Stacked Bar chart
plt.bar(df['batsman'],df['2017'],label='2017', color='green')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016', color='yellow')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016'] + df['2017']),label='2015', color='red')

plt.legend()
plt.show()
```

# 3D Plots

## 3D Scatter

```python

fig = plt.figure(figsize=(7,10))
ax = plt.subplot(projection='3d')
ax.scatter3D(df['avg'], df['runs'], df['strike_rate'], color='green')
ax.set_title('Avs Vs SR vs Runs')
ax.set_xlabel(' Avg')
ax.set_ylabel(' Runs')
ax.set_zlabel('Strike Rate')
```

## 3D Line Plot

```python
fig = plt.figure(figsize=(7,7))
ax = fig.add_subplot(projection='3d')
x = [2,4,5,6]
y = [5,6,10,12]
z = [3,1,10,5]

ax.scatter3D(x,y,z)
ax.plot3D(x,y,z, color='green')
plt.show()
```

```python

fig = plt.figure(figsize=(10,5))

# First 3D subplot
ax1 = fig.add_subplot(1, 2, 1, projection='3d')
ax1.scatter3D(df['avg'], df['runs'], df['strike_rate'], color='green')
ax1.set_title('Avs Vs SR vs Runs')
ax1.set_xlabel(' Avg')
ax1.set_ylabel(' Runs')
ax1.set_zlabel('Strike Rate')

# Second 3D subplot
ax2 = fig.add_subplot(1, 2, 2, projection='3d')
ax2.scatter3D(x, y, z, color='red')
ax2.plot3D(x, y, z, color='red')
ax2.set_title("Plot 2")

plt.tight_layout()
plt.subplots_adjust(wspace=0.2)
plt.show()
```

## Surface Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx, yy = np.meshgrid(x,y)

z = xx ** 2 + yy ** 2


fig, ax = plt.subplots(1,1, subplot_kw={'projection':'3d'}, figsize=(7,7))
c = ax.plot_surface(xx,yy,z, cmap='jet')
fig.colorbar(c)
```

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx, yy = np.meshgrid(x,y)
z = xx**4 + yy**4


# fig = plt.figure(figsize=(12,8))

# ax = plt.subplot(projection='3d')

fig, ax = plt.subplots(1,1, subplot_kw={'projection':'3d'}, figsize=(12,8))

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

## Heatmap

```python
delivery = pd.read_csv('data/deliveries.csv')
delivery.head(3)
```

```python
mask = (delivery['ballnumber'].isin([1,2,3,4,5,6])) & (delivery['batsman_run'] == 6)
d = delivery[mask]
d
info = d.groupby(['overs', 'ballnumber']).aggregate(
    batsman = ('batsman_run', 'count')
).unstack()
```

```python
plt.imshow(info)
```

```python
info2 = d.pivot_table(index='overs', columns=['ballnumber'], values='batsman_run', aggfunc='count')
plt.figure(figsize=(10,10))
plt.imshow(info2)
plt.xticks(range(0,6), range(1,7))
plt.yticks(range(0,20), range(1,21))
plt.colorbar()
```
