```python

# from a pandas dataframe
batsman = pd.read_csv('data/sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'],batsman['V Kohli'], color='red', linestyle='solid', marker='D', markersize=10, label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'], color='green', linestyle='dashed', marker='o', linewidth=2, label='Rohit')
plt.title('Rohit vs Virat Career comparisons')
plt.xlabel('Years')
plt.ylabel('Runs')
plt.legend(loc='upper left')
plt.show()
```

# Scatter Plots

```python
# plt.scatter on pandas data
df = pd.read_csv('data/batter.csv')
df = df.head(50)
df.head(10)
```

```python

plt.figure(figsize=(10, 10))
plt.scatter(df['avg'], df['strike_rate'], s=df['runs']/100, marker='+', color='green')
plt.title('Strike Rate vs Avg')
plt.ylabel('Strike Rate')
plt.xlabel('Avg')

for _, row in df.iterrows():
    plt.text(row['avg'], row['strike_rate'], row['batter'], fontsize=8)
```

# Bar Chart

```python
# color and label
df = pd.read_csv('data/batsman_season_record.csv')
df
```

```python
plt.bar(df.index - 0.2, df['2015'], width=0.2,color='yellow', label='2015')
plt.bar(df.index, df['2016'],width=0.2, color='red', label='2016')
plt.bar(df.index+0.2, df['2017'],width=0.2, color='green', label='2017')
plt.legend()
plt.title('Batsman run over yers')
plt.xlabel('Batsman')
plt.ylabel('Runs')
plt.xticks(df.index, df['batsman'], rotation='vertical')
plt.show()
```

# Stacked Bar

```python
# Stacked Bar chart
plt.bar(df['batsman'],df['2017'],label='2017', color='green')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016', color='yellow')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016'] + df['2017']),label='2015', color='red')

plt.legend()
plt.show()
```
